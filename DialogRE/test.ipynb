```python
def language_metrics(lang, metrics_list):
    """Helper function to load language metrics"""
    metrics = {}
    for m in metrics_list:
        metrics[m] = f'./data/dict/{lang}/{m}.txt'
    return metrics

# Define emotion categories
emotion_labels = [
    'positive', 'negative', 'neutral', 'rude',
    'catastrophizing', 'dichotomous-reasoning', 'disqualifying-positive', 
    'emotional-reasoning', 'fortune-telling', 'labeling', 'magnification', 
    'mental-filtering', 'mindreading', 'overgeneralizing', 'personalizing', 
    'should-statement'
]

# Initialize emotion detector
emotion_detector = TextMetrics(language_metrics('en', emotion_labels), debug=False)
sentiment_detector = PygentsSentiment(
    './data/dict/en/positive.txt',
    './data/dict/en/negative.txt', 
    debug=False
)

print("Emotion detection systems initialized")
```
